
# Databricks SQL

## Explanation
Databricks SQL is a serverless data warehouse solution on Databricks that enables users to run SQL queries on data stored in the Lakehouse. It provides a familiar SQL interface for data exploration, analytics, and reporting, leveraging the scalability and performance of Databricks.

## Features
- **Serverless Architecture**: No infrastructure management required.
- **Lakehouse Integration**: Direct access to Delta Lake tables and unified data storage.
- **Advanced Query Optimization**: Uses Photon engine for high-performance SQL execution.
- **Built-in Visualization**: Create dashboards and visualizations directly from SQL queries.
- **Security & Governance**: Supports fine-grained access controls and auditing.
- **Collaboration**: Share queries, dashboards, and results with team members.
- **Scheduling**: Automate query execution and dashboard refreshes.

## Uses
- **Data Exploration**: Analyze and profile data using SQL.
- **Business Intelligence**: Build dashboards and reports for business users.
- **Ad-hoc Analysis**: Run quick queries to answer business questions.
- **ETL Operations**: Transform and prepare data for downstream applications.
- **Data Governance**: Monitor and audit data access and usage.


## SQL Warehouse

### What is it?
A **SQL Warehouse** is the compute engine in Databricks that executes your SQL queries. It's the processing power behind Databricks SQL.

### Key Characteristics

* **Serverless or Provisioned**: Choose between fully managed serverless (instant startup) or provisioned clusters (more control)
* **Photon Engine**: Built-in query acceleration for fast analytics performance
* **Auto-scaling**: Automatically adjusts compute resources based on query demand
* **Pay-per-use**: With serverless, you only pay for active query time
* **Multi-cluster**: Can scale to multiple clusters to handle concurrent users

### Types

| Type | Best For | Startup Time |
| --- | --- | --- |
| **Serverless** | BI dashboards, ad-hoc queries, instant access | Instant |
| **Pro** | Predictable workloads, concurrent users | ~2 minutes |
| **Classic** | Basic SQL workloads | ~2 minutes |

### Simple Analogy
Think of it like a car engine:
* **Data** = the road you're driving on
* **SQL Query** = your destination 
* **SQL Warehouse** = the engine that gets you there

Without a SQL Warehouse, your queries can't run - just like a car needs an engine to move.

# SQL Interview Cheat Sheet

## 1. Stored Procedure

**Definition:** A stored procedure is a precompiled group of SQL statements stored in the database. It is mainly used to **perform an operation**.

```sql
CREATE PROCEDURE GetEmployees()
BEGIN
    SELECT * FROM employees;
END;
```

Call it:

```sql
CALL GetEmployees();
```

**Interview point:** A procedure can perform `INSERT`, `UPDATE`, `DELETE`, and `SELECT`.

---

## 2. Function

**Definition:** A function is a stored program that **returns a value**.

```sql
CREATE FUNCTION GetBonus(salary DECIMAL)
RETURNS DECIMAL
RETURN salary * 0.10;
```

Use:

```sql
SELECT employee_name, GetBonus(salary)
FROM employees;
```

### Procedure vs Function

| Procedure | Function |
|---|---|
| Mainly performs an operation | Mainly calculates/returns a value |
| May or may not return data | Must return a value |
| Called using `CALL` | Can be used in SQL expressions |
| Can contain DML | Usually used for calculations |

**Interview answer:**  
> A procedure is used to perform a task, while a function is generally used to calculate and return a value.

---

## 3. Trigger

**Definition:** A trigger is automatically executed when a specific database event occurs.

Common events:

- `INSERT`
- `UPDATE`
- `DELETE`

Example:

```sql
CREATE TRIGGER after_employee_insert
AFTER INSERT ON employees
FOR EACH ROW
INSERT INTO employee_log(employee_id, action)
VALUES (NEW.id, 'INSERTED');
```

Whenever a new employee is inserted, the trigger automatically records it.

**Interview question: Can we call a trigger manually?**

No. A trigger is automatically fired when its associated event occurs.

---

## 4. Exception Handling

**Definition:** Exception handling is used to handle runtime errors without abruptly terminating the program/procedure.

Example:

```sql
BEGIN
    DECLARE CONTINUE HANDLER FOR SQLEXCEPTION
    BEGIN
        SELECT 'Error occurred';
    END;

    INSERT INTO employees VALUES (1, 'Rahul');
END;
```

Common exceptions/errors:

- Duplicate key
- Constraint violation
- Invalid data
- Division by zero
- SQL execution failure

**Interview answer:**

> Exception handling allows us to catch and handle runtime database errors gracefully.

---

## 5. Parameters

There are three important parameter types in stored procedures.

### IN

Input parameter.

```sql
CREATE PROCEDURE GetEmployee(IN emp_id INT)
BEGIN
    SELECT *
    FROM employees
    WHERE id = emp_id;
END;
```

```sql
CALL GetEmployee(10);
```

### OUT

Returns a value from the procedure.

```sql
CREATE PROCEDURE GetCount(OUT total INT)
BEGIN
    SELECT COUNT(*) INTO total
    FROM employees;
END;
```

### INOUT

Used for **both input and output**.

```sql
CREATE PROCEDURE UpdateValue(INOUT x INT)
BEGIN
    SET x = x + 10;
END;
```

### Easy way to remember

- **IN → Give data**
- **OUT → Get data**
- **INOUT → Give + Get**

---

## 6. Lexical Elements

**Definition:** Lexical elements are the basic building blocks recognized by the SQL parser.

Example:

```sql
SELECT employee_name
FROM employees
WHERE salary > 50000;
```

Here:

- `SELECT`, `FROM`, `WHERE` → **Keywords**
- `employee_name`, `employees`, `salary` → **Identifiers**
- `50000` → **Literal/value**
- `>` → **Operator**
- `;` → **Delimiter**

**Interview definition:**

> Lexical elements are the smallest meaningful components of SQL statements, such as keywords, identifiers, literals, operators, and delimiters.

# 7. Joins

Suppose we have:

### employees

| id | name | dept_id |
|---|---|---|
| 1 | A | 10 |
| 2 | B | 20 |
| 3 | C | 30 |

### departments

| dept_id | department |
|---|---|
| 10 | IT |
| 20 | HR |
| 40 | Finance |

---

## INNER JOIN

Returns only matching records.

```sql
SELECT e.name, d.department
FROM employees e
INNER JOIN departments d
ON e.dept_id = d.dept_id;
```

Result:

```text
A  IT
B  HR
```

**Remember:** INNER = matching rows only.

---

## LEFT JOIN

Returns **all rows from the left table** and matching rows from the right table.

```sql
SELECT e.name, d.department
FROM employees e
LEFT JOIN departments d
ON e.dept_id = d.dept_id;
```

Result:

```text
A  IT
B  HR
C  NULL
```

**Remember:** LEFT = everything from left.

---

## RIGHT JOIN

Returns all rows from the right table.

```sql
SELECT e.name, d.department
FROM employees e
RIGHT JOIN departments d
ON e.dept_id = d.dept_id;
```

Finance appears even though no employee belongs to it.

---

## FULL OUTER JOIN

Returns **all records from both tables**.

```sql
SELECT e.name, d.department
FROM employees e
FULL OUTER JOIN departments d
ON e.dept_id = d.dept_id;
```

**Remember:**

```text
INNER → matching
LEFT  → all left
RIGHT → all right
FULL  → everything
```

---

## CROSS JOIN

Produces every possible combination.

```sql
SELECT e.name, d.department
FROM employees e
CROSS JOIN departments d;
```

If there are 3 employees and 3 departments:

```text
3 × 3 = 9 rows
```

---

## SELF JOIN

A table joins with itself.

Example: employee-manager relationship.

```sql
SELECT e.name AS employee,
       m.name AS manager
FROM employees e
LEFT JOIN employees m
ON e.manager_id = m.id;
```

# 8. Cursor

**Definition:** A cursor allows us to process query results **one row at a time**.

Example:

```sql
DECLARE emp_cursor CURSOR FOR
SELECT name FROM employees;
```

Typical cursor flow:

```text
DECLARE cursor
      ↓
OPEN cursor
      ↓
FETCH row
      ↓
Process row
      ↓
FETCH next row
      ↓
CLOSE cursor
```

Example:

```sql
OPEN emp_cursor;

FETCH emp_cursor INTO emp_name;

CLOSE emp_cursor;
```

**Interview point:**

> A cursor processes rows individually, but set-based SQL is generally preferred because it is usually more efficient.

---

# 9. Window Function

**Definition:** A window function performs calculations across related rows **without collapsing the rows**.

Example:

```sql
SELECT
    employee_name,
    department,
    salary,
    AVG(salary) OVER(PARTITION BY department) AS avg_dept_salary
FROM employees;
```

Each employee still appears as a separate row, but each gets the department's average salary.

---

## ROW_NUMBER()

Assigns a unique sequential number.

```sql
SELECT
    employee_name,
    salary,
    ROW_NUMBER() OVER(ORDER BY salary DESC) AS rn
FROM employees;
```

Example:

```text
A  100000  1
B   90000  2
C   80000  3
```

---

## RANK()

Same values receive the same rank, with gaps.

```sql
SELECT
    employee_name,
    salary,
    RANK() OVER(ORDER BY salary DESC) AS rnk
FROM employees;
```

For:

```text
100
100
90
```

Ranks:

```text
1
1
3
```

---

## DENSE_RANK()

Same values receive the same rank, but **no gaps**.

```text
100 → 1
100 → 1
90  → 2
```

### ROW_NUMBER vs RANK vs DENSE_RANK

| Function | Duplicate values | Gap? |
|---|---|---|
| ROW_NUMBER | Different numbers | No |
| RANK | Same rank | Yes |
| DENSE_RANK | Same rank | No |

---

# 10. PARTITION BY

`PARTITION BY` divides rows into groups for a window function.

```sql
SELECT
    employee_name,
    department,
    salary,
    RANK() OVER(
        PARTITION BY department
        ORDER BY salary DESC
    ) AS dept_rank
FROM employees;
```

This ranks employees **within each department**.

---

# 11. GROUP BY vs Window Function

### GROUP BY

Reduces rows.

```sql
SELECT department, AVG(salary)
FROM employees
GROUP BY department;
```

Result:

```text
IT  80000
HR  70000
```

### Window Function

Keeps individual rows.

```sql
SELECT
    employee_name,
    department,
    salary,
    AVG(salary) OVER(PARTITION BY department) AS avg_salary
FROM employees;
```

Result:

```text
A  IT  80000  80000
B  IT  70000  80000
C  IT  90000  80000
```

**Interview answer:**

> `GROUP BY` combines rows into groups; window functions calculate across groups while retaining individual rows.

# 12. Common SQL Clauses

Typical order you **write** SQL:

```text
SELECT
FROM
WHERE
GROUP BY
HAVING
ORDER BY
LIMIT
```

Example:

```sql
SELECT department, AVG(salary) AS avg_salary
FROM employees
WHERE salary > 30000
GROUP BY department
HAVING AVG(salary) > 50000
ORDER BY avg_salary DESC
LIMIT 5;
```

### WHERE vs HAVING

**WHERE** → filters rows **before grouping**

```sql
WHERE salary > 30000
```

**HAVING** → filters groups **after GROUP BY**

```sql
HAVING AVG(salary) > 50000
```

# 18. CASE

SQL equivalent of if/else.

```sql
SELECT
    name,
    salary,
    CASE
        WHEN salary >= 100000 THEN 'HIGH'
        WHEN salary >= 50000 THEN 'MEDIUM'
        ELSE 'LOW'
    END AS salary_category
FROM employees;
```

# 19. Subquery

A query inside another query.

```sql
SELECT *
FROM employees
WHERE salary > (
    SELECT AVG(salary)
    FROM employees
);
```

Meaning:

> Find employees earning more than the average salary.

---

# 20. CTE

CTE = **Common Table Expression**.

It makes complex queries easier to read and can be referenced like a temporary result set within the statement.

```sql
WITH high_salary AS (
    SELECT *
    FROM employees
    WHERE salary > 50000
)
SELECT *
FROM high_salary;
```

# 21. UNION vs UNION ALL

### UNION

Combines results and removes duplicates.

```sql
SELECT name FROM employees
UNION
SELECT name FROM managers;
```

### UNION ALL

Combines results but keeps duplicates.

```sql
SELECT name FROM employees
UNION ALL
SELECT name FROM managers;
```

**Interview point:** `UNION ALL` is generally faster because it does not perform duplicate elimination.

# 23. Index

**Definition:** An index improves data retrieval speed.

```sql
CREATE INDEX idx_employee_name
ON employees(name);
```

Without an index, the database may need to scan many rows.

With an appropriate index, the database can locate matching rows more efficiently.

**Trade-off:** Indexes improve reads but can make `INSERT`, `UPDATE`, and `DELETE` more expensive and consume storage.

# 24. NULL

`NULL` means **missing/unknown value**, not zero and not an empty string.

Wrong:

```sql
WHERE manager_id = NULL
```

Correct:

```sql
WHERE manager_id IS NULL
```

```sql
WHERE manager_id IS NOT NULL
```

# 25. COALESCE

Returns the first non-NULL value.

```sql
SELECT COALESCE(phone, 'No Phone')
FROM employees;
```

```text
No Phone
```

# 26. Aggregate Functions

Aggregate functions perform calculations on multiple rows.

Common functions:

```sql
COUNT()
SUM()
AVG()
MIN()
MAX()
```

Example:

```sql
SELECT
    COUNT(*) AS total_employees,
    SUM(salary) AS total_salary,
    AVG(salary) AS average_salary,
    MIN(salary) AS minimum_salary,
    MAX(salary) AS maximum_salary
FROM employees;
```

# 30. Transactions

A transaction is a group of operations treated as one logical unit.

Important commands:

- `COMMIT` → permanently save changes
- `ROLLBACK` → undo changes
- `SAVEPOINT` → create a point to roll back to

Example:

```sql
START TRANSACTION;

UPDATE accounts
SET balance = balance - 1000
WHERE id = 1;

UPDATE accounts
SET balance = balance + 1000
WHERE id = 2;

COMMIT;
```

If something goes wrong:

```sql
ROLLBACK;
```

---

# 31. ACID Properties

Transactions follow ACID principles:

| Property | Meaning |
|---|---|
| Atomicity | All operations succeed or all fail |
| Consistency | Data remains valid |
| Isolation | Concurrent transactions don't improperly interfere |
| Durability | Committed changes survive failures |

**Easy interview answer:**

> ACID ensures reliable and consistent transaction processing.
