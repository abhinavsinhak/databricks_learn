### COPY INTO vs Auto Loader

| Feature         | COPY INTO                          | Auto Loader                        |
|-----------------|------------------------------------|------------------------------------|
| Use Case        | Batch ingestion                    | Streaming/continuous ingestion     |
| Trigger         | Manual or scheduled                | Automatic (triggered by new files) |
| File Detection  | Scans all files each run           | Tracks new files efficiently       |
| Schema Evolution| Limited                            | Supported                          |
| Complexity      | Simple SQL command                 | Requires setup with Structured Streaming |
| Performance     | May reprocess files                | Scalable, avoids reprocessing      |

**COPY INTO** is best for simple, batch loads.  
**Auto Loader** is ideal for scalable, incremental, and streaming ingestion.

```python
%sql
CREATE TABLE data_engineer.bronze.copyinto(
    order_id int,
    customer_id int,
    order_date date,
    order_amount double,
    order_status string
)
```

```python
%sql
COPY INTO data_engineer.bronze.copyinto
FROM "/Volumes/data_engineer/bronze/autovol/copyinto/"
FILEFORMAT = PARQUET
FORMAT_OPTIONS ('mergeSchema' = 'true')
COPY_OPTIONS ('mergeSchema' = 'true')
```
